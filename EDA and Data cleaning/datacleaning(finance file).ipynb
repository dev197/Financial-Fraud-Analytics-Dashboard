## Cleaning data and changing data types

```python
df['transaction_type'] =df['transaction_type'].astype('category')
```

```python
df['channel'].unique()
```

```python
df['channel'] =df['channel'].str.strip()
```

```python
df['channel'] = df['channel'].replace({"M@bile App": "Mobile App"})
```

```python
df['channel'].unique()
```

```python
df['channel'] =df['channel'].astype('category')
```

```python
df['merchant_category'] = df['merchant_category'].astype('category')
```

```python
df['currency'].unique()
```

```python
df['currency'] = df['currency'].replace({"inr":"INR","inR":"INR"})
```

```python
df['currency'].unique()
```

```python
df['currency'] =df['currency'].astype('category')
```

```python
df['is_fraud'] =df['is_fraud'].astype('category')
```

```python
df['transaction_status'] = df['transaction_status'].astype('category')
```

```python
df['transaction_date'] = pd.to_datetime(df['transaction_date'], format='%d-%m-%Y')
```

```python
## checking for missing values
df.isnull().sum()
```

```python
df = df.dropna(subset=['fee_amount'])
```

```python
df.shape
```

## Checking duplicate values

```python
df.duplicated().sum()
```

```python
## dropping duplicate values
df = df.drop_duplicates()
```

```python
df = df.reset_index(drop=True)
```

```python
df.shape
```

## Adding Year and month columns

```python
df['transaction_year'] = df['transaction_date'].dt.year
```

```python
df['transaction_month'] = df['transaction_date'].dt.month
```

```python
df.to_csv("../Dataset/Cleaned/Cleaned_finance_transactions.csv")
```
