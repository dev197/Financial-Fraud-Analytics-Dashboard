```python
## Fixing the Date type 
df['date_of_birth'] = pd.to_datetime(df['date_of_birth'], format='%d-%m-%Y')
df['join_date'] = pd.to_datetime(df['join_date'], format='%d-%m-%Y')
```

```python
## Renaming column
df = df.rename(columns={'fisrt_name':'first_name'})
```

```python
df.info()
```

```python
cols = ['first_name', 'second_name', 'gender', 'city', 'occupation', 'customer_segment']
for cols in cols:
    print(f'column {cols} {df[cols].unique()}')
```

## Converting data type to category

```python
cols = ['gender', 'city', 'state', 'occupation', 'customer_segment']
for cols in cols:
    df[cols] = df[cols].astype('category')
```
